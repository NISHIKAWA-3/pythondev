```python
words_list =[]

alpha = dict.fromkeys(list("abcdefghijklmnopqrstuvwxyz"),0)

while True:
    words_input = input("英単語を入力してください： ")
    print(f"英単語を入力してください: {words_input} ")
    if words_input =="":
        break
    words_list.append(words_input) 
words_list.sort()
  
print(words_list)

for word in words_list:
    print("今のword:", word)
    for letter in word:
        print("  今のletter:", letter)
        if letter in alpha:
            alpha[letter]+=1

for k, v in alpha.items():
    if v >= 1:
        print(f"{k}が{v}個ありました") 
 


```

```python
mylist = []
alphabet = dict.fromkeys("\
abcdefghijklmnopqrstuvwxyz", 0)

while True:
    word = input('英単語を入力してください')
    if word == '':
        break
    mylist.append(word)
mylist.sort()
print(mylist)

for w in mylist:
    for c in w:
        try:
            alphabet[c] += 1
        except KeyError:
            pass

for k, v in alphabet.items():
    print(f'{k}が{v}個ありました')
```
