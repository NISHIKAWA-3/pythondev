```python
class Person:
    # コンストラクタ（初期化メソッド）
    def __init__(self, name, age):
        self.name = name  # インスタンス変数
        self.age = age

    # メソッド
    def greet(self):
        print(f"こんにちは、私は{self.name}、{self.age}歳です。")

# クラスからインスタンスを作成
p1 = Person("太郎", 20)
p2 = Person("花子", 18)

# メソッド呼び出し
p1.greet()  # こんにちは、私は太郎、20歳です。
p2.greet()  # こんにちは、私は花子、18歳です。  「クラスは設計図」「インスタンスは実体」「selfは自分自身」
```

```python
class Student(Person):  # Person を継承
    def __init__(self, name, age, student_id):
        super().__init__(name, age)  # 親クラスのコンストラクタ呼び出し
        self.student_id = student_id

    def greet(self):
        print(f"私は学生です。名前は{self.name}、IDは{self.student_id}です。")
```

```python
class MyClass:
    class_var = 0            # クラス変数

    def __init__(self, v):
        self.inst_var = v    # インスタンス変数

a = MyClass(1)
b = MyClass(2)

print(a.class_var, b.class_var, MyClass.class_var)
# 出力: 0 0 0  ← 共通のクラス変数を見ている

MyClass.class_var = 5
print(a.class_var, b.class_var)
# 出力: 5 5  ← クラスを書き換えると全インスタンスから見える値が変わる

a.class_var = 10
print(a.class_var, b.class_var, MyClass.class_var)
# 出力: 10 5 5 ← `a.class_var = 10` はインスタンス属性を作る（シャドーイング）

del a.class_var
print(a.class_var)
# 出力: 5 ← インスタンス属性を削除すると再度クラス属性が参照される
```

```python
class Bag:
    items = []   # ← クラス変数（ミュータブル）

a = Bag()
b = Bag()
a.items.append('apple')
print(b.items)   # ['apple'] ← 期待外：a と b で共有される！
```

```python
class Baseball_Team:
    def __init__(self, name, win, lose, draw): #__int__の中でメンバ変数name...を定義
        self.name = name #指定された名前をメンバ変数nameへ代入するよう定義。インスタンスメンバ変数。
        self.win = win
        self.lose = lose
        self.draw = draw
        self.win_rate = self.calc_win_rate()

    
    def calc_win_rate(self):
        result = (self.win/ (self.win + self.lose))
        return round(result, 3)

    def show_team_result(self):
        team_result = (f"{self.name:<10}{self.win:<8}{self.lose:<8}{self.draw:<6} {self.calc_win_rate():<8}")
        print(team_result)
        
```

```python
archers = Baseball_Team("Archers", 77, 64, 2)
meteors = Baseball_Team("Meteors", 71, 69, 3)	
cheetahs = Baseball_Team("Cheetahs", 69, 68, 6)
sharks = Baseball_Team("Sharks", 70, 70, 3)
pythons = Baseball_Team("Pythons", 68, 73, 2)
owls = Baseball_Team("Owls", 59, 82, 2)

print(f"{'team':<10}{'win':<8}{'lose':<8}{'draw':<7}{'rate':<8}")

archers.show_team_result()
meteors.show_team_result()
cheetahs.show_team_result()
sharks.show_team_result()
pythons.show_team_result()
owls.show_team_result()


```

```python
class PurchasedItem:
    def __init__(self, item_name):  #初期化メソッド。self→これから作るオブジェクト自身。item_name→オブジェクトを作るときに外から与える情報
        self.item_name = item_name #「外から受け取ったitem_name（例: 'a cup of coffee'）を、このオブジェクトが持つitem_nameという変数に保存してください」
                                   #self.変数名と書くことで、そのオブジェクト固有のデータを保存しておくことができる。（属性（アトリビュート））
    def get_item_name(self):  #このPurchasedItemオブジェクトができる「アクション（振る舞い）」を定義
        return self.item_name  #return self.item_nameは、「このオブジェクトが持っているitem_name属性の値を返してください」という意味

#設計図を使って、実際に「モノ（オブジェクト）」を作り出す。このプロセスを「インスタンス化」という
item = PurchasedItem('a cup of coffee')  #PurchasedItemクラスの設計図を元に、新しいオブジェクトを1つ作成。'a cup of coffee'という文字列が、先ほど説明した__init__メソッドのitem_name引数に渡される
                                         #「item_nameが'a cup of coffee'である」という情報を持ったオブジェクトが生成され、itemという変数に代入

item_name = item.get_item_name()  #先ほど作成したitemオブジェクトに対して、「あなたが持っているget_item_nameメソッドを実行してください！」と命令
                                  #get_item_nameメソッドは、self.item_nameを返すように作られていたので、itemオブジェクトのitem_nameは'a cup of coffee'なので、
                                  #このメソッドは'a cup of coffee'という文字列を返す。その結果が、item_nameという新しい変数に格納
                            

print('I bought', item_name, 'at a convenience store.')
```
