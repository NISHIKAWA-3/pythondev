```python
def fizz_buzz(x):
    if x % 15 ==0:
        result = "FizzBuzz"
    elif x % 3 ==0:
        result = "Fizz" 
    elif x % 5 ==0:
        result = "Buzz"
    else:
        result = x
    return result    
r1 = fizz_buzz(3)
r2 = fizz_buzz(5)
r3 = fizz_buzz(30)
r4 = fizz_buzz(49)
print(r1, r2, r3, r4)       
```

```python
def fib(i):
    if i == 0:
        return 0
    if i == 1:
        return 1
    return fib(i-1) + fib(i-2)
    
```

```python
fib(9)
```
