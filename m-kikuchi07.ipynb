```python
def fizz_buzz(x):
    if x % 15 ==0:
        result = "FizzBuzz"
    elif x % 3 ==0:
        result = "Fizz" 
    elif x % 5 ==0:
        result = "Buzz"
    else:
        result = x
    return result    
r1 = fizz_buzz(3)
r2 = fizz_buzz(5)
r3 = fizz_buzz(30)
r4 = fizz_buzz(49)
print(r1, r2, r3, r4)       
```

```python
def fib(i):
    if i == 0:
        return 0
    if i == 1:
        return 1
    return fib(i-1) + fib(i-2)
    
```

```python
fib(9)
```

```python
def add(a, b):  # a, bが引数
    return a + b  # 戻り値としてa + bを返す

result = add(3, 5)  # 3と5を引数として渡す
print(result)  # 戻り値を表示
```

```python
def calculate_area(width, height):  # 幅と高さを引数に取る
    area = width * height
    return area  # 面積を戻り値として返す

area = calculate_area(4, 5)  # 幅4、高さ5を渡す
print(f"面積は{area}平方メートルです")
```

```python
def add_numbers(num1, num2):

  #2つの数値の合計を計算する関数
  return num1 + num2

def calculate_average(total, count):

  #合計と要素数から平均値を計算する関数
  return total / count

# 2つの数値の合計を計算
total = add_numbers(10, 20)

# 合計と要素数から平均値を計算
average = calculate_average(total, 2)

print(average)
```
