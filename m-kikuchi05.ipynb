```python
char = "e" # char: **charater**から
char_count = ta.count(char)
if char_count: # `if char_count != 0:`と同等
    print(f"文字「{char}」はリスト{ta}に{char_count}回含まれます。")
else:
    print(f"文字「{char}」はリスト{ta}に含まれません。")

char = "x"
char_count = ta.count(char)
if char_count:
    print(f"文字「{char}」はリスト{ta}に{char_count}回含まれます。")
else:
    print(f"文字「{char}」はリスト{ta}に含まれません。")
```
