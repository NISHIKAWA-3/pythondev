```python
price = int(input("税抜き価格を入力してください: "))
print(f"税抜き価格を入力してください:{price} ")
including_tax = round(price * 1.1)

if including_tax >= 2000:
    print("送料は無料です")
    print("送料込みの価格は", including_tax, "円です。")
else:
    print("送料として350円かかります")
    print("送料込みの価格は",including_tax + 350, "円です。")
```

```python
count = 0

while True:
    score = int(input("点数を入力してください: "))
    print(f"点数を入力してください:{score} ")
    if score == -1:
        break
    count += 1

print(f"入力回数: {count}")


```

```python
total = 0
count = 0

while True:
    score = int(input("点数を入力してください: "))
    print(f"点数を入力してください:{score} ")
    if score == -1:
        break
    total += score
    count += 1
    

score_average = total / count
round_score_ave=round(score_average, 1)

print(f"{count}人のテスト平均点は{round_score_ave}点です")

```
