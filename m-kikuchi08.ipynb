```python
class Suit:
    def __init__(self, color, size, brand):
        self.color = color
        self.size = size
        self.brand = brand

    def __call__(self, occasion):
        return f"この{self.color}のスーツ（サイズ：{self.size}、ブランド：{self.brand}）は、{occasion}にぴったりです。"

# クラスのインスタンス化
my_suit = Suit(color="ピンク", size="L", brand="サンプルブランド")

# インスタンスを関数のように呼び出す
result = my_suit("パーティー")
result
```

```python
class Person:
    # コンストラクタ（初期化メソッド）
    def __init__(self, name, age):
        self.name = name  # インスタンス変数
        self.age = age

    # メソッド
    def greet(self):
        print(f"こんにちは、私は{self.name}、{self.age}歳です。")

# クラスからインスタンスを作成
p1 = Person("太郎", 20)
p2 = Person("花子", 18)

# メソッド呼び出し
p1.greet()  # こんにちは、私は太郎、20歳です。
p2.greet()  # こんにちは、私は花子、18歳です。
        
```

```python
class Student(Person):  # Person を継承
    def __init__(self, name, age, student_id):
        super().__init__(name, age)  # 親クラスのコンストラクタ呼び出し
        self.student_id = student_id

    def greet(self):
        print(f"私は学生です。名前は{self.name}、IDは{self.student_id}です。")
```

```python
class MyClass:
    class_var = 0            # クラス変数

    def __init__(self, v):
        self.inst_var = v    # インスタンス変数

a = MyClass(1)
b = MyClass(2)

print(a.class_var, b.class_var, MyClass.class_var)
# 出力: 0 0 0  ← 共通のクラス変数を見ている

MyClass.class_var = 5
print(a.class_var, b.class_var)
# 出力: 5 5  ← クラスを書き換えると全インスタンスから見える値が変わる

a.class_var = 10
print(a.class_var, b.class_var, MyClass.class_var)
# 出力: 10 5 5 ← `a.class_var = 10` はインスタンス属性を作る（シャドーイング）

del a.class_var
print(a.class_var)
# 出力: 5 ← インスタンス属性を削除すると再度クラス属性が参照される
```

```python
from dataclasses import dataclass 
from typing import ClassVar

@dataclass
class Cat:
    name: str = "猫"
    voice: ClassVar[str] = "ニャー"

    def description(self):
        print(f"{self.name}です。鳴き声は{self.voice}です。")
```

```python
cat1 = Cat()
cat1.description()
cat2 = Cat("タマ")
cat2.description()
```
