```python
#matplotlib.pyplotのインポート,matplotlib.pyplot をモジュールとしてインポート
import matplotlib.pyplot as plt #←を実行することで、matplotlib.pyplot モジュールを plt という名前で利用できます。pyplot を plt と名づけるのは、pandasをpd、numpyをnpとするのと同様、慣習によるもの
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("./sample_data/california_housing_train.csv")
df
```

```python
y = df.median_income # .valuesとすればndarrayが取得できるが、そのままで支障ない。median_incomeのデータを抽出
y
```

```python
#グラフを作成する,plt.plot() を実行.引数には表示したいデータの配列（ここでは y）を指定
plt.plot(y)
```

横軸が配列のインデックス、縦軸は y の値（median_income）で作成したグラフが表示される

```python
#補足：作成したグラフを表示する,[<matplotlib.lines.Line2D at 0xccccccccc>] のような、余計なテキストが表示されるのを避けるためplt.show() とつけ加える
plt.plot(y)
plt.show()
```
