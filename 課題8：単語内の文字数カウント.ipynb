```python
words_list =[]

alpha = dict.fromkeys(list("abcdefghijklmnopqrstuvwxyz"),0)

while True:
    words_input = input("英単語を入力してください： ")
    print(f"英単語を入力してください: {words_input} ")
    if words_input =="":
        break
    words_list.append(words_input) 
words_list.sort()
  
print(f"入力した単語{words_list}")
      

for word in words_list:
    for letter in word:
        if letter in alpha:
            alpha[letter]+=1

for k, v in alpha.items():
    if v >= 1:
        print(f"{k}が{v}個ありました") 
```
